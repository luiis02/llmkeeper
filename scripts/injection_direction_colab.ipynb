```python
import os
from google.colab import userdata

# El token es opcional: solo hace falta si la descarga de embeddinggemma-2 devuelve 401/403
try:
    os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
except Exception:
    pass
os.environ["INJECTION_EMBED_MODEL"] = "google/embeddinggemma-2"
os.makedirs("scripts", exist_ok=True)
print("Token cargado:", bool(os.environ.get("HF_TOKEN")))
```

Sube `scripts/injection_direction.py` al entorno (panel de archivos a la izquierda, o con el selector de abajo).

```python
from google.colab import files

uploaded = files.upload()  # selecciona injection_direction.py
for name, data in uploaded.items():
    with open(os.path.join("scripts", name), "wb") as f:
        f.write(data)
print(os.listdir("scripts"))
```

```python
# Ejecuta el pipeline. Los logs salen aquí y en scripts/injection_direction.log
!python scripts/injection_direction.py
```

## Notas

- Tanto `SentenceTransformer` como `GemmaEncoder` se ejecutan en CPU en el script actual. Un runtime con GPU no acelera nada hasta que el encoder use `cuda`.
